```python
# Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Uploading the dataset
df = pd.read_csv("/content/Used_Car_Price_Prediction.csv")
```

```python
df.drop_duplicates(inplace = True)
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df['body_type'] = df['body_type'].fillna(df['body_type'].mode().iloc[0])
```

```python
df['transmission'] = df['transmission'].fillna(df['transmission'].mode().iloc[0])
```

```python
df['registered_city'] = df['registered_city'].fillna(df['registered_city'].mode().iloc[0])
```

```python
df['registered_state'] = df['registered_state'].fillna(df['registered_state'].mode().iloc[0])
```

```python
df['source'] = df['source'].fillna(df['source'].mode().iloc[0])
```

```python
df['car_availability'] = df['car_availability'].fillna(df['car_availability'].mode().iloc[0])
```

```python
df['original_price'] = df['original_price'].fillna(df['original_price'].mean())
```

```python
df['car_rating'] = df['car_rating'].fillna(df['car_rating'].mode().iloc[0])
```

```python
df['ad_created_on'] = df['ad_created_on'].fillna(df['ad_created_on'].mode().iloc[0])
```

```python
df['fitness_certificate'] = df['fitness_certificate'].fillna(df['fitness_certificate'].mode().iloc[0])
```

```python
df.isnull().sum()
```

```python
#Outline
plt.figure(figsize = (15,5))
sns.boxplot(df[['yr_mfr', 'kms_run', 'sale_price', 'times_viewed', 'total_owners', 'broker_quote', 'original_price',
                'emi_starts_from', 'booking_down_pymnt']])
plt.title("Outlier Checking")
plt.show()
```

```python
#Outliner Treatment
for column in ['yr_mfr', 'kms_run', 'sale_price', 'times_viewed', 'total_owners', 'broker_quote', 'original_price',
                'emi_starts_from', 'booking_down_pymnt']:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = np.where(df[column]>upper_bound,upper_bound,np.where(df[column]<lower_bound,lower_bound,df[column]))
```

```python
#Change of the Outliner
plt.figure(figsize = (15,5))
sns.boxplot(df[['yr_mfr', 'kms_run', 'sale_price', 'times_viewed', 'total_owners', 'broker_quote', 'original_price',
                'emi_starts_from', 'booking_down_pymnt']])
plt.title("Outlier Checking")
plt.show()
```
